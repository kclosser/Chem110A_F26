## Basic Plotting

```python
# import libraries
# we will use the matplotlib library for doing plots and numpy for doing numerical fitting.
import matplotlib.pyplot as plt
import numpy.polynomial.polynomial as poly
```

```python

# define points
x_vals = [1, 2, 3, 5, 7]
y_vals = [1.10, 2.28, 3.37, 5.11, 6.76]

# calculate a linear fit
b, m = poly.polyfit(y_vals, x_vals, 1)
print(f"Slope = {m:4f}\nIntercept = {b:4f}")

# generate y values to plot line
line = []
for x in x_vals:
    line.append(m*x+b)

# plot the points and line
plt.plot(x_vals, y_vals, "o", label="Practice data")
plt.plot(x_vals, line, "gray", label=f"Linear fit: y = {m:.6}*x + {b:.3}")

plt.xlabel('x values')
plt.ylabel('y values')
plt.legend()
plt.show()
```
